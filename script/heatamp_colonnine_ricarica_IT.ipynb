```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/csv/rete_ricarica_veicoli_elettrici.csv')


# Metodo 3: usando value_counts() (più diretto)
conteggi = df['regione'].value_counts()
print(conteggi)
print(sum(conteggi))

# Metodo 1: Heatmap verticale con seaborn
plt.figure(figsize=(10, 8))
sns.heatmap(conteggi.values.reshape(-1, 1), 
            annot=True,  # mostra i numeri
            fmt='d',     # formato intero
            cmap='YlOrRd',  # scala di colori
            yticklabels=conteggi.index,
            cbar_kws={'label': 'Numero di stazioni'})
plt.xlabel('')
plt.ylabel('Regione')
plt.title('Stazioni di ricarica per Regione')
plt.tight_layout()
```
