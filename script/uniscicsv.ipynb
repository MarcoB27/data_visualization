```python
import pandas as pd

files = [
    '../data/csv/nuove_auto_ibride_europa.csv',  # file1 [file:11]
    '../data/csv/nuove_auto_plugin_europa.csv', # file2 [file:12]
    '../data/csv/nuove_auto_plugin_europa2.csv'
]

dfs = []
for i, file in enumerate(files, 1):
    print(f"Carico {file}...")
    df = pd.read_csv(file)
    df = df[df['mot_nrg'] == 'DIE'].copy()  # Solo diesel
    df['OBS_VALUE'] = pd.to_numeric(df['OBS_VALUE'], errors='coerce')
    df['file'] = f'file{i}'  # Traccia origine
    dfs.append(df)
    print(f"  Shape: {df.shape}, Anni: {sorted(df['TIME_PERIOD'].unique())}")

# Unione multi-file con groupby somma
df_unito = pd.concat(dfs, ignore_index=True)
df_finale = df_unito.groupby(['geo', 'TIME_PERIOD'])['OBS_VALUE'].sum().reset_index()
df_finale = df_finale.rename(columns={'geo': 'paese', 'TIME_PERIOD': 'anno', 'OBS_VALUE': 'diesel_totali'})
df_finale = df_finale.dropna().sort_values(['paese', 'anno'])

# Salva
output = 'nuove_ibride_europa.csv'
df_finale.to_csv(output, index=False)
print(f"\nSalvato: {output} ({len(df_finale)} righe)")
print("\nPrime righe:")
print(df_finale.head(10))
print("\nPaesi:", df_finale['paese'].nunique())
print("Anni:", sorted(df_finale['anno'].unique()))
```
